# Reshaping Array
- the size of the initial array must match the size of the reshaped array

# Reshaping

```python
array = np.arange(1,10)
print("Array originally: " , array)
print(array.ndim)
print('\n')

ar_res = array.reshape(3,3)
print('Reshaped Array:\n' , ar_res)
print(ar_res.ndim)
```

```python
array = np.arange(1,13)
print("Array originally: " , array)
print(array.ndim)
print('\n')
ar_res = array.reshape(6,2)
print('Reshaped Array:\n' , ar_res)
print(ar_res.ndim)
```

```python
trans_arr = np.transpose(ar_res)
print('Transpose of Array:\n' , trans_arr)
```

##### Iteration over array

```python
a = np.arange(0, 45, 5)
a
```

```python
a = a.reshape(3,3)
a
```

```python
for x in np.nditer(a):
    print(x)
```

```python
print(a)
#printing row wise
for x in np.nditer(a, order = 'C'):
    print(x)

# printing column wise
print('\n')
for x in np.nditer(a, order = 'F'):
    print(x)
```
